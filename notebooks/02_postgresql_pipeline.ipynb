## I. Giới thiệu
Mục tiêu: Xây dựng nền tảng dữ liệu cho toàn bộ dự án AI.
Pipeline: CSV → Import PostgreSQL → Join / Aggregation → Flat Table → Notebook 03


## II. Tạo Database và table
Thực hiện trên pgAdmin4:
```sql
CREATE DATABASE climate_db;

CREATE TABLE city_temp (
    dt DATE,
    AverageTemperature FLOAT,
    AverageTemperatureUncertainty FLOAT,
    City VARCHAR(255),
    Country VARCHAR(255),
    Latitude VARCHAR(50),
    Longitude VARCHAR(50)
);
```


## III. Import dữ liệu
Sử dụng lệnh COPY trong pgAdmin4:
```sql
COPY city_temp(dt, AverageTemperature, AverageTemperatureUncertainty, City, Country, Latitude, Longitude)
FROM 'C:\GlobalLandTemperaturesByMajorCity.csv'
DELIMITER ','
CSV HEADER;
```


## IV. Tối ưu Database
```sql
CREATE INDEX idx_city ON city_temp(City);
CREATE INDEX idx_dt ON city_temp(dt);
```


## V & VI. Join dữ liệu & Tạo Flat Table / View
Với bài toán này, bảng `city_temp` đã tương đối phẳng. Ta có thể tạo một view chỉ chứa dữ liệu đã filter (ví dụ: thành phố Tokyo):
```sql
CREATE VIEW vw_tokyo_temp AS
SELECT * FROM city_temp WHERE City = 'Tokyo';
```
